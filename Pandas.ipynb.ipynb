```python
#1. Create a DataFrame named student_df from a dictionary containing the columns Name, Age, and City for four students
data = {
    'Name' : ['Rahul' , 'Dev','Ayush','Kashish'],
    'Age' : [20,21,24,23],
    'City' : ['Bangkok','Hodal','Palwal','Delhi']
}
student_df = pd.DataFrame(data)
print(student_df)
```

```python
#2. Display the complete student_df DataFrame.
print(student_df)
```

```python
#3. Write code to display the column names of student_df.
print(student_df.columns)
```

```python
#4. Retrieve the Name column from student_df.
print(student_df['Name'])
```

```python
#5. Retrieve the City column from student_df using square-bracket notation.
print(student_df['City'])
```

```python
#6. Create a new DataFrame from student_df containing only the columns Name and City.
new_df = student_df[['Name','City']]
print(new_df)

```
