== checks whether two values are equal, while is checks whether two variables refer to the same object.

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
```

```python
x = 100
y = 100

print(x is y)
print(x == y)

p = 1000
q = 1000

print(p is q)
print(p == q)
```
